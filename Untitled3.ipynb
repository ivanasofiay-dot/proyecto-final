```python
import matplotlib.pyplot as plt
import csv
```

```python
Mes=[]
Pozo_A=[]
Pozo_B=[]
Pozo_C=[]
Pozo_D=[]
Total_mensual=[]
Total_acumulado=[]
```

```python
produccion='producciones.csv'
with open(produccion,'r') as archivo:
    lectura= csv.reader(archivo)
    header=next(lectura)
    for datos in lectura:
        Mes.append(str(datos[0]))
        Pozo_A.append(float(datos[1]))
        Pozo_B.append(float(datos[2]))
        Pozo_C.append(float(datos[3]))
        Pozo_D.append(float(datos[4]))
        Total_mensual.append(float(datos[5]))
        Total_acumulado.append(float(datos[6]))
```

```python
plt.bar(Mes,Pozo_A, color='pink')
plt.title('Producción mensual del Pozo A')
plt.xlabel('Mes')
plt.ylabel('Producción')
plt.show()

plt.bar(Mes,Pozo_B, color='indigo')
plt.title('Producción mensual del Pozo B')
plt.xlabel('Mes')
plt.ylabel('Producción')
plt.show()

plt.bar(Mes,Pozo_C, color='mediumpurple')
plt.title('Producción mensual del Pozo C')
plt.xlabel('Mes')
plt.ylabel('Producción')
plt.show()

plt.bar(Mes,Pozo_D, color='blueviolet')
plt.title('Producción mensual del Pozo D')
plt.xlabel('Mes')
plt.ylabel('Producción')
plt.show()
```

```python
plt.plot(Mes,Total_acumulado, color='red')
plt.title('Producción mensual acumulada')
plt.xlabel('Mes')
plt.ylabel('Producción')
plt.show()
```

```python
total_PA=sum(Pozo_A)
total_PB=sum(Pozo_B)
total_PC=sum(Pozo_C)
total_PD=sum(Pozo_D)

totales=[total_PA,total_PB,total_PC,total_PD]
etiquetas=['Pozo_A','Pozo_B','Pozo_C','Pozo_D']

plt.pie(totales, labels=etiquetas, autopct='%1.1F%%', colors= ['greenyellow', 'lightcoral', 'skyblue','mediumpurple'])
plt.title('Aportación de cada pozo')
plt.show()
```
